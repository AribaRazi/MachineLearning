```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
df.head()
```

```python
df['species'].unique()
```

```python
from sklearn.model_selection import train_test_split

X= df.drop('species',axis=1)
y= df['species']
```

```python
X
```

```python
# train test split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
# creating a model
from sklearn.neighbors import KNeighborsClassifier
```

```python
model_knn=KNeighborsClassifier(n_neighbors=13)

model_knn.fit(X_train,y_train)

```

```python
y_pred=model_knn.predict(X_test)
print(model_knn.score(X_test,y_test))
```

```python
# now using SVM
```

## SVM

```python
from sklearn.svm import SVC
```

```python
model_svm=SVC(gamma='auto')
```

```python
model_svm.fit(X_train,y_train)
model_svm.score(X_test,y_test)
```

```python
# now keep chaning the values of the parameters like C kernel etc
```

### now lets use the GRID SEARCH CV

```python
from sklearn.model_selection import GridSearchCV

# lets tune the hyperparameters ie creating grids
classifier = GridSearchCV((model_svm),{
    'C':[1,10,20,30],
    'kernel' : ['liear','rbf'],
    # cv for cross validation
},cv=5,return_train_score=False)
```

```python
classifier.fit(X,y)
```

```python
classifier.cv_results_
```

```python
results=pd.DataFrame(classifier.cv_results_)
```

```python
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

# RANDOM SEARCH CV

```python
from sklearn.model_selection import RandomizedSearchCV

classifier_r = RandomizedSearchCV((model_svm),{
    'C':[1,10,20,30],
    'kernel' : ['liear','rbf'],
    # cv for cross validation
},n_iter=4,cv=5,return_train_score=False)
```

```python
classifier_r.fit(X,y)
```

```python
results_cv=pd.DataFrame(classifier_r.cv_results_)
```

```python
results_cv
```

```python
results_cv[['param_C','param_kernel','mean_test_score']]
```
